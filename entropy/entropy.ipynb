### Deriving Entropy

Entropy is, given a probabilistic distribution, on average what is the minimal number of bits to describe it.

#### 0. Base example
fliping a coin:

A -> 50%

B -> 50%

For each flip, we need 1 bit, so we can differentate A and B.

>H = 1 bit/flip

---

#### 1. Start with 10 coin flips
What if:

A -> 10%

B -> 90%

Since most of the time we just get B, we definitely need less than 1 bit/flip to differentiate them!

For example, we can just **remember A's position**!

Let's say we only do 10 flips (in theory it should be infinite), and we know that **exactly one A** appears in 10 flips, how many possible sequences are there?

The A can appear in any of the 10 positions:

```text
BBBBBBBBBA
BBBBBBBBAB
BBBBBBBABB
...
ABBBBBBBBB
```

There are exactly:

$$
\binom{10}{1}=10
$$

possible sequences.

To distinguish among 10 possibilities, we need:

$$
\lceil \log_2 10 \rceil = 4 \text{ bits}
$$

Why log2? 

1 bit can distinguish 2 possibilities.

$k$ bits can distinguish \(2^k\) possibilities.

So H here is:
>4/10 = 0.4 bits/flip

---

#### 2. What if there are two A's?
What if:

A -> 20%

B -> 80%

Still 10 flips and now suppose exactly **two A's** appear in 10 flips.

Examples:

```text
AABBBBBBBB
ABABBBBBBB
ABBABBBBBB
...
BBBBBBBBAA
```

We need to choose 2 positions out of 10:

$$
\binom{10}{2}=\frac{10!}{2!8!}=45
$$

That's 45 possible sequences to remember!

Therefore, we need:

$$
\lceil \log_2 45 \rceil = 6 \text{ bits}
$$

Notice how quickly the number of possible sequences grows.

The information needed to describe a sequence is determined by **how many different sequences are possible**.

So H here is:
>6/10 = 0.6 bits/flip

---

#### 3. One last example

Suppose:

- P(A) = 1%
- P(B) = 99%

We flip independently 10,000 times.

A typical sequence contains approximately:

- 100 A's
- 9,900 B's

Assume, for now, that we know the sequence contains exactly 100 A's.

You know the drill now.

The number of possible sequences is:

$$
N=\binom{10000}{100}
$$

Thus, the minimum number of bits needed to identify one of these sequences is approximately:

$$
\log_2 \binom{10000}{100}
$$

So H here is:
>803/10000 = 0.0803 bits/flip

---

#### 4. Entropy emerges naturally

So far, we've made one convenient assumption: 
>we know exactly how many A's appear in each sequence.

Let's remove that assumption.

As the sequence gets longer, the observed frequency of A converges to its true probability $p$.

Suppose we perform $n$ independent flips, with:

$$
P(A)=p,\qquad P(B)=1-p
$$

A typical long sequence contains approximately:

- $np$ occurrences of A
- $n(1-p)$ occurrences of B

The number of sequences with exactly $np$ occurrences of A is (a.k.a what we need to remember):

$$
N=\binom{n}{np}
$$

The number of bits needed to distinguish these sequences is approximately:

$$
L=\log_2\binom{n}{np}
$$

Using Stirling's approximation, for large $n$:

$$
\log_2\binom{n}{np}
\approx
-n\left[p\log_2p+(1-p)\log_2(1-p)\right]
$$

Divide by $n$ to obtain the number of bits per flip:

$$
\frac{L}{n}
\approx
-p\log_2p-(1-p)\log_2(1-p)
$$

And there it is:

$$
\boxed{H(p)=-p\log_2p-(1-p)\log_2(1-p)}
$$

**Shannon entropy emerges directly from counting possible sequences.**
