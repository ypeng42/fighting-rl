### Question: how cpu uses MMU?

Ex. when a user program calls:

```c
p[1000000] = 'A';
```

How is MMU involved? CPU doesn't know process id, it's a software concept!

### Answer: everytime CPU switch process, kernel tells CPU to switch to that process's page table

```text
Running Process A:
    satp → Page Table A

Running Process B:
    satp → Page Table B
```

Setting satp configures the MMU. Memory access uses the MMU automatically. 

1 4kb root page table per process is enough (level-1, level-2 table could be many). Kernel just needs to track process -> root page table mapping (level-1, level-2 table location is like a tree, auto-tracked).

---

xv6 uses identity mapping, see `kvmmake(void)`, so that before and after MMU enabled, CPU can use the same address.
