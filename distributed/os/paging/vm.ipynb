| Function | Data Direction |
|---|---|
| `copyout()` | Kernel → User |
| `copyin()` | User → Kernel |
| `copyinstr()` | User → Kernel, copies a null-terminated (`\0`) string |

why need such? Ex. copyout()

User call read(), kernel will read file from disk and copy the content to the buffer send by user.

So copyout() needs to translate VA to PA.

Why kernel has MMU turned on but it can use physical address?

Beacuse kernel uses Identity Mapping! it maps VA to the same PA.

---

Every process's user page table needs to contain trampoline mapping.

```c
pagetable_t
proc_pagetable(struct proc *p)
{
  pagetable_t pagetable = uvmcreate();

  mappages(
    pagetable,
    TRAMPOLINE,
    PGSIZE,
    (uint64)trampoline,
    PTE_R | PTE_X
  );

  // Other mappings omitted
  return pagetable;
}
```
