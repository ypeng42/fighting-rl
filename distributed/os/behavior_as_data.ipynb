# Behavior as Data: A Unifying Idea in Computer Science

## 1. The Core Insight

**A program is an order. A CPU is a worker. An operating system is a factory manager.**

Computers are not fundamentally different from factories:

| Factory | Computer |
|---|---|
| Order | Program |
| Worker | CPU |
| Instructions | Machine Code |
| Work in progress | Process |
| Workbench | Registers |
| Workspace | Memory / Stack |
| Factory Manager | OS Scheduler |

Code is simply a precise description of work to be done.

This analogy explains why **time-sharing** works: a CPU can switch between different tasks, just as a worker can switch between orders.

The important distinction is:

- **Program:** A description of what needs to be done.
- **Process:** A program with its own execution state and resources.

Conceptually:

\[
\text{Process} = \text{Program} + \text{Execution State}
\]

Two processes can execute the same program while having completely different execution states.

The CPU does not fundamentally care which process it is executing. It simply executes instructions.

The OS manages which process gets to use the CPU.

---

## 2. The Big Idea: Turn Behavior into Data

One of the most powerful ideas in computer science is:

> **Turn "doing something" into "a description of doing something" so another system can inspect, manage, transform, or execute it.**

This idea is closely related to **Reification**: turning an abstract concept, such as behavior or execution, into something that can be represented and manipulated.

It appears everywhere:

| System | What Becomes Data? | What Does This Enable? |
|---|---|---|
| Lisp | Code → S-expressions | Generate and manipulate code |
| OS | Execution → Process State | Pause, resume, schedule |
| Spark | Computation → DAG | Analyze dependencies, schedule and optimize |
| SQL | Query → Relational Expression | Choose execution plans |
| Compiler | Program → Intermediate Representation | Analyze and optimize code |
| Kubernetes | Desired State → API Objects | Reconcile actual state with desired state |

These systems share a general architecture:

```text
       Description / State
                |
                v
      Manager / Interpreter
                |
                v
            Execution
```

The key insight:

**When behavior is directly executed, we can observe its results. When behavior is represented as data, we gain the opportunity to reason about and control its execution.**

However, these representations are not identical:

- Lisp represents program structure.
- An OS maintains execution state.
- Spark represents computation and dependencies.

They share a design principle, not necessarily the same implementation.

## 3. How Much Information Must We Represent?

A deeper question emerges:

**How much of a behavior must be represented for another system to pause, resume, reschedule, or optimize it?**

The answer depends on what we want the manager to do.

Consider:

```python
a = 2 + 3
b = 7 * 4
print(a + b)
```

### Level 1: Pause and Resume

Suppose we execute the first instruction and pause.

To resume correctly, we must know:

1. **Where** execution stopped.
2. **What state** must be preserved.

Conceptually:

\[
\text{Resumable Task} = \text{Code} + \text{State}
\]

This is central to OS process management.

**The OS does not need to understand what the computation means to pause and resume it.**

### Level 2: Reschedule Execution

Suppose we want to execute `b` before `a`, or execute them in parallel.

Now we need to understand their **dependencies**.

```text
     a = 2 + 3       b = 7 * 4
          \             /
           \           /
            v         v
             print(a+b)
```

Since `a` and `b` are independent, they can be computed in either order.

But `print(a + b)` depends on both.

Conceptually:

\[
\text{Schedulable Task} = \text{Operations} + \text{Dependencies}
\]

This is a central idea behind Spark DAG scheduling.

Correct scheduling must also respect side effects and other execution constraints.

### Level 3: Optimize Execution

Suppose we want to transform:

```python
a = 2 + 3
b = 7 * 4
print(a + b)
```

Into:

```python
print(33)
```

Now dependencies alone are insufficient.

We need to understand the **semantics** of the operations and which transformations preserve observable behavior.

Conceptually:

\[
\text{Optimizable Task} = \text{Operations} + \text{Semantics} + \text{Constraints}
\]

This is a central idea behind compiler optimization and database query optimization.

---

## 4. A General Principle: Representation Determines Control

| Desired Capability | Required Information |
|---|---|
| Execute | What to execute next |
| Pause / Resume | Execution position and necessary state |
| Reschedule | Dependencies and execution constraints |
| Optimize | Semantics, valid transformations, and costs |

This gives us a powerful design principle:

> **The amount of control a system can exercise over a behavior depends on the information exposed by its representation.**

A representation does not need to expose everything.

It needs to expose enough information for the operations we want to support.

This is the essence of good abstraction:

**Expose what is necessary. Hide what is irrelevant.**

An important limitation is that external side effects require special care. Saving and restoring CPU state does not automatically undo a network request or make a repeated bank transfer safe.

---

## 5. Why OS, Spark, and Compilers Are Different

These systems manage computations at different levels of understanding.

**OS Scheduler**

"I don't need to understand what you're computing. I just need to save and restore your execution context."

**Spark Scheduler**

"I need to understand which tasks depend on which others, so I can distribute and schedule their execution."

**Compiler Optimizer**

"I need to understand which transformations preserve the program's observable behavior, so I can produce a more efficient implementation."

Their differences arise partly from **how much information they require about the behavior they manage**.

More information can enable more powerful transformations, but it also introduces additional complexity.
