# Why Separate `.text`, `.data`, and `.bss`?

## 1. `.text` vs `.data`: Memory Protection

**Problem:** Code and data need different memory permissions.

- **`.text` (Code):** Read + Execute, but not Write.
- **`.data` (Variables):** Read + Write, but not Execute.

**Solution:** Place code and data in separate memory pages so the CPU can enforce different permissions.

**Benefit:** Prevents accidental code modification and reduces security risks.

## 2. `.bss`: Avoid Storing Unnecessary Zeros

**Problem:** A program may need a large amount of zero-initialized memory.

```c
char buffer[1024 * 1024 * 1024]; // 1 GB
```

Storing 1 GB of zeros in the executable would waste disk space and disk I/O.

**Solution:** Instead of storing the zeros:

- **On disk:** Record how much zero-initialized memory is needed.
- **At load time:** Allocate and zero that memory.
- **At runtime:** The program gets the same 1 GB zero-initialized array.

**Benefit:** Smaller executable files and less disk I/O.

## 3. Where Does ELF Fit In?

ELF is a **standardized contract between the linker and the loader**.

It describes:

- Which parts of a program should be loaded into memory.
- Where they should be placed.
- How much memory they require.
- What permissions they need.
- Where execution should begin (Entry Point).

**Key Takeaway:** Separating code, data, and zero-initialized memory solves real problems. ELF provides a standard way to describe how the loader should prepare them.
