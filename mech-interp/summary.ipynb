
## Mechanistic Interpretability

Reverse-engineering a trained neural network.

### A. Representation

What does the model know? Where is the information stored?

- Linear Probes
- Sparse Autoencoders (SAEs)
- Activation Oracles

### B. Computation

How does the model use this information?

- Induction Heads
- IOI Circuits
- SAE Circuits

### A+B -> C. Causal Validation

Modify internal states and test whether the outputs change as predicted.

- Activation Patching
- Ablation
- Steering
