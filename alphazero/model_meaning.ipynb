## Environment Model in Reinforcement Learning

In reinforcement learning, an **environment model** typically consists of two core functions:

$$
\boxed{s' = T(s,a)}
$$

$$
\boxed{r = R(s,a,s')}
$$

- **\(T\) — Transition Model:** Given the current state \(s\) and an action \(a\), what will the next state \(s'\) be?
- **\(R\) — Reward Model:** What reward does the agent receive after taking an action?

For **Go**, these functions represent:

- **\(T\):** The rules of the game, including placing stones, capturing stones, switching turns, and determining legal moves.
- **\(R\):** The rewards associated with winning or losing at the end of the game.

These functions constitute the **environment model used by AlphaZero**. AlphaZero does not need a neural network to learn these rules because they are already known.

## Why Is DQN Model-Free While AlphaZero Is Model-Based?

Both DQN and AlphaZero have access to game simulators, but they use them differently.

### DQN (Model-Free)

$$
s \xrightarrow{\text{environment interaction}} s' \xrightarrow{\text{learning}} Q_\theta
$$

During training, DQN interacts with the environment to collect experience and update its Q-network. During action selection, it typically chooses actions directly based on predicted Q-values, without using the simulator to explore hypothetical future states.

### AlphaZero (Model-Based)

$$
s \xrightarrow{\text{internal simulation using }T} s' \xrightarrow{\text{NN evaluation}} v
$$

During action selection, AlphaZero actively uses the environment model to simulate future states and expand its Monte Carlo Tree Search (MCTS). Its neural network provides policy priors and value estimates, while the game rules determine how the simulated states evolve.

### Key Takeaway

**DQN uses the environment to generate training data, whereas AlphaZero additionally uses the environment model for planning.**

The distinction between model-free and model-based reinforcement learning is not whether the agent knows the game rules, but **whether it uses a model of the environment to simulate hypothetical future states and guide decision-making**.
