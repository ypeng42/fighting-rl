# Root Dirichlet Noise

**Core idea:** Randomly boost a few actions at the root of MCTS to encourage exploration.

### Intuition: Restaurant Analogy

- **ε-greedy:** Occasionally pick a random restaurant and eat there without checking.
- **Dirichlet Noise + MCTS:** Occasionally pick a restaurant you wouldn't normally consider, research its reviews and menu, then decide whether to eat there.

### Formula

$$
P'=(1-\epsilon)P+\epsilon\eta
$$

- $P$: Original NN policy.
- $\eta$: Random probability distribution sampled from a Dirichlet distribution.
- $\epsilon$: How much noise to mix into the policy.

### Alpha (α): Noise Concentration

- **Smaller α → Spikier noise:** A few random actions receive most of the noise.
- **Larger α → More uniform noise:** Noise is spread more evenly across actions.

A useful heuristic:

$$
\boxed{\alpha\approx\frac{10}{K}}
$$

Where $K$ is the approximate number of legal actions.

**Why?** With hundreds of possible actions, spreading noise too thinly may have little impact. Concentrating it on a few randomly selected actions gives MCTS a meaningful opportunity to investigate them.

**Key takeaway:** Randomize *what to explore*, not necessarily *what to play*.
