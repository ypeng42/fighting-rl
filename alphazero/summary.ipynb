# AlphaZero — Key Insights

**AlphaZero = RL-generated targets + Supervised Learning**
- Value Head → MSE against actual game outcomes
- Policy Head → Cross-Entropy against MCTS results
- RL generates better targets; SL learns to imitate them.

**Value = Judgment | Policy = Intuition | MCTS = Thinking**
- Value Head = Compress experience into judgment.
- Policy Head = Distill expensive search into intuition.
- MCTS = Spend computation to improve decisions.
