# GRPO, RL, and the Emergence of Chain of Thought

## Key Takeaways

1. **GRPO did not invent CoT.** It is an optimization algorithm that determines how rewards are used to update the model.
2. **RFT (Rejection Sampling Fine-Tuning) and REINFORCE can also reinforce CoT.** GRPO's specific gradient weighting is not necessary.
3. **CoT emergence and gradient weighting are two different but related questions:**
   - **What behavior does the model learn?** (CoT)
   - **How does training select and reinforce that behavior?** (Optimization)

## 1. Where Does CoT Come From?

A pretrained LLM has already seen mathematical proofs, step-by-step explanations, and reasoning patterns.

Even before RL, it may generate two types of responses:

- **Strategy A:** Directly output the answer.
- **Strategy B:** Reason step by step, verify intermediate results, then output the answer (CoT).

**RL does not need to invent CoT from scratch. It can discover and reinforce reasoning behaviors that already have some probability under the pretrained model.**

Importantly, DeepSeek-R1-Zero was not completely free of reasoning guidance:

- Its prompt template instructed the model to put reasoning inside `<think>` tags and the final answer inside `<answer>` tags.
- Training also included a **format reward**.
- However, it did not use supervised fine-tuning on human-written CoT examples before RL.

## 2. How Does RL Make CoT More Common?

Consider a hypothetical pretrained model:

| Strategy | Generation Probability | Accuracy |
|---|---|---|
| A: Direct answer | 90% | 10% |
| B: CoT | 10% | 90% |

Out of 100 generated responses:

- Strategy A: 90 attempts → 9 correct.
- Strategy B: 10 attempts → 9 correct.

Although CoT initially represents only **10% of all responses, it accounts for 50% of correct responses**.

If we use RFT to train on successful responses, CoT becomes more likely to be generated.

REINFORCE and GRPO can create similar selection pressure, but they differ in how they weight and update sampled responses.

**CoT can gradually evolve from a rare output pattern into the model's preferred problem-solving strategy.**

## Final Insight

No particular GRPO gradient-weighting formula explicitly encodes *"generate CoT"* or *"learn to self-reflect."*

**CoT is a complete output trajectory. RL uses rewards to change the probability of generating that trajectory and its constituent tokens.**

The deeper question is not *"Why does GRPO create reasoning?"* but:

**Why can optimizing outcome rewards on a pretrained LLM reinforce increasingly sophisticated reasoning behaviors?**
