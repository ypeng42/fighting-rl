## Online RL: Footsies vs. LLM PPO-based RLHF

The core training loop is nearly identical in both systems.

| Footsies RL | LLM PPO-based RLHF |
|---|---|
| Agent (policy) | LLM being trained (policy) |
| Initial game state | User prompt |
| Action (move, attack, etc.) | Generating one token |
| Complete game (trajectory) | Complete response (trajectory) |
| Game outcome / reward | Reward Model score |
| PPO updates the agent | PPO updates the LLM |

### The Online RL Loop

**Policy generates experience → Evaluator provides reward → Optimizer updates policy → Updated policy generates new experience.**

The key idea is that the policy continuously generates fresh training data as it improves.

### Important Distinction

- **Footsies:** The agent interacts with an environment and an opponent. The environment responds to its actions.
- **LLM RLHF:** The LLM generates responses, and a Reward Model evaluates them. The Reward Model acts as a scorer, not an opponent.

In both cases, the policy learns from its own newly generated experience. This is what makes the training **online**.

## 1 year after InstructGPT -> DPO

In terms of the training mechanism, DPO is much closer to Pretraining / SFT than to PPO.

Your mental model is accurate, with one important addition: DPO changes not only the loss function but also the training data format—from individual target responses to preference pairs.

|                                  | Pretraining / SFT                    | DPO                               | PPO                                                 |
| -------------------------------- | ------------------------------------ | --------------------------------- | --------------------------------------------------- |
| Training data                    | Fixed text or demonstrated responses | Fixed preference pairs            | Rollouts generated by the current policy            |
| Loss function                    | Cross-entropy loss                   | Preference loss                   | Policy gradient surrogate loss                      |
| Optimization method              | Gradient descent                     | Gradient descent                  | Gradient ascent / descent on the negative objective |
| Requires online data generation? | No                                   | No, for standard DPO              | Yes, for standard PPO                               |
| Requires a Reward Model?         | No                                   | No explicit Reward Model required | Yes, in classical PPO-based RLHF                    |


From an implementation perspective:

\\[ \boxed{\text{DPO} \approx \text{SFT-style Training Loop} + \text{Preference Loss}} \\]

This describes the implementation mechanism, not the underlying theoretical objective.
