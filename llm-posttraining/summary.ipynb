# Understanding LLM Post-Training Terminology

## 1. A Framework for Classifying Training Methods

Instead of treating SFT, DPO, PPO, and GRPO as sequential stages, we can classify training methods along three independent dimensions.

### Dimension 1: Feedback Source

**Who or what determines whether a model's output is good?**

- Human preferences
- AI judges
- Unit tests or rule-based verifiers
- Environment rewards

### Dimension 2: Optimization Method

**How are the model's parameters updated?**

- SFT
- DPO
- PPO
- GRPO
- REINFORCE variants

### Dimension 3: Data Collection

**Does the model continuously generate new attempts during training?**

- **Offline:** Training on a fixed dataset.
- **Online:** Collecting fresh rollouts as training progresses.

## 2. What Does "RLHF" Actually Mean?

Suppose an engineer says:

> We used RLHF to improve the model.

This statement leaves three important questions unanswered:

1. **Where does the feedback come from?**
2. **What objective or optimizer did you use?**
3. **Were the training samples collected offline or online?**

Answering these questions tells us much more about the actual training procedure than the term *RLHF* alone.

## 3. A Mental Model of LLM Training

- **Pretraining** learns from text.
- **SFT** learns from demonstrations.
- **DPO** learns from preference comparisons.
- **Online RL** learns from feedback on newly generated attempts.

These methods describe different learning mechanisms, not necessarily different capabilities or sequential training stages.

## 4. Terminology Cheat Sheet

| Category | Terms | What They Describe |
|---|---|---|
| Feedback source | RLHF, RLAIF, RLVR | Where the learning signal comes from |
| Optimization algorithm | PPO, GRPO | How the policy is optimized |
| Training objective or method | SFT, DPO | How demonstrations or preferences are used |
| Data collection | Offline, Online | How training samples are collected |
| Desired outcome | Reasoning, Alignment | What capabilities or behaviors we want to improve |

**Key takeaway:** RLHF, RLAIF, and RLVR describe the feedback used for learning; PPO and GRPO describe policy optimization algorithms; reasoning and alignment describe the capabilities or behaviors we aim to improve.

These categories are not mutually exclusive. For example, GRPO can be used with human-preference rewards (RLHF) or verifiable rewards (RLVR), and both can involve online reinforcement learning.
