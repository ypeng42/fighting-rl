# Q-Table vs. Neural Network

## Core Insight

The fundamental trade-off between a Q-table and a neural network are:

- Parameter Independence vs. Parameter Sharing
- Discrete Representation vs. Continuous Function Approximation

Strictly speaking, **parameter sharing is the mechanism; generalization is a potential benefit.**

tabular methods often have stronger convergence guarantees in small, finite MDPs, while neural networks become attractive in large or continuous state spaces where maintaining independent parameters is impractical.


## Comparison

| Dimension | Q-Table | Neural Network |
|---|---|---|
| Parameter Independence | High | Low (shared parameters) |
| Parameter Sharing | None across entries | Yes |
| Generalization | No inherent generalization to unseen states | Can generalize to unseen states |
| Update Interference | Minimal across entries | Possible |
| Sample Efficiency | Requires experience for individual entries | Can reuse learned patterns across states |
| Approximation Bias | None with complete tabular representation | Possible |
| Scalability | Limited by state-action space size | Better suited to large or continuous spaces |
| Convergence Guarantees | Available under standard tabular assumptions | No general guarantee for nonlinear RL |

## The Fundamental Trade-off

$$
\text{Parameter Independence}
\quad\longleftrightarrow\quad
\text{Parameter Sharing / Generalization}
$$


Parameter sharing introduces an **inductive bias**: the assumption that different state-action pairs contain useful common structure.

- If this assumption is correct, parameter sharing can improve **generalization and sample efficiency**.
- If this assumption is incorrect, parameter sharing can introduce **approximation error** and update interference.

## A Spectrum of Function Approximation

Q-tables and neural networks are not the only options.

Intermediate approaches include:

- Linear Function Approximation
- Tile Coding
- Radial Basis Functions (RBF)
- Shared Embeddings

These approaches provide different ways to balance parameter independence and knowledge sharing.
