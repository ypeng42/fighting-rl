Therefore, **TD error serves two purposes**:

- Training the Critic: As an error signal for updating value predictions (ex. TD(0)).
- Updating the Actor: As an estimate of the advantage of a particular action (Policy Gradient).

Advantage is the **expected value** of the TD error. TD error is one sample.

### A. How do we coordinate Actor and Critic?

**1. Critic Optimality Before Actor Update**
- Framework: Idealized analysis
- Reference: Sutton et al. (1999), Theorem 3

↓  

**2. Two-Time-Scale Actor-Critic**
- Framework: Concurrent learning with convergence analysis
- Reference: Konda & Tsitsiklis (1999)

↓

**3. A2C / A3C / PPO / APPO**
- Framework: Practical deep RL approaches
- Key idea: Different update schedules and optimization objectives, without generally enforcing Konda's assumptions

### B. How do we estimate Advantage?

**1. Monte Carlo Returns**
- Use sampled future rewards

↓

**2. TD / n-step Returns**
- Introduce bootstrapping to reduce variance

↓

**3. GAE (Generalized Advantage Estimation)**
- Combine multi-step TD residuals to tune the bias–variance tradeoff
