### variance vs. bias summary


| Method                 | Signal Used by Actor                    | Gradient Bias                                            | Variance                                 |
| ---------------------- | --------------------------------------- | -------------------------------------------------------- | ---------------------------------------- |
| REINFORCE              | \\(G_t\\)                               | 0                                                        | Typically high                           |
| REINFORCE + Baseline   | \\(G_t - b(s_t)\\)                      | 0                                                        | Typically lower with a suitable baseline |
| 1999 Compatible Critic | \\(f_w(s_t,a_t)\\)                      | 0 when both conditions are satisfied                     | Potentially reduced                      |
| Actor–Critic + TD(0)   | \\(r_t + \gamma V(s\_{t+1}) - V(s_t)\\) | May be biased when the value approximation is inaccurate | Typically lower                          |
| Actor–Critic + GAE     | Weighted sum of multiple TD errors      | Depends on \\(\lambda\\), value estimation errors, etc.  | Tunable                                  |

### policy gradient components

| Component           | Responsibility                                                | Common Methods                      |
| ------------------- | ------------------------------------------------------------- | ----------------------------------- |
| Actor               | Updates the policy \\(\pi\_\theta\\)                          | Policy Gradient                     |
| Critic              | Learns the value function \\(V\_\phi\\) or \\(Q\_\phi\\)      | TD Learning, Monte Carlo (MC), etc. |
| Advantage Estimator | Constructs the training signal \\(\hat{A}\_t\\) for the Actor | MC − Baseline, TD Error, GAE        |

---

**Why do most Actor–Critic algorithms still train the Critic primarily using Value Prediction Loss, rather than directly minimizing Actor Gradient Estimation Error?**

Value Learning can construct MC, TD, or λ-return targets from trajectories and then perform regression directly.

Another practical reason is that a trained $V(s)$ can be reused for bootstrapping across many states, rollouts, and policy updates, rather than serving only a single Actor gradient estimate.

Therefore, **Value MSE is a surrogate loss that is easy to construct, easy to optimize, and generally useful, but is not fully aligned with the Actor's ultimate objective.**

This does not mean that Critic objectives cannot be designed to directly reduce gradient variance or bias. Such objectives are simply more complex and typically require additional estimation.

<br>

- REINFORCE + Baseline: Does not require an accurate Critic. It has an **unbiased expected policy gradient**, but the **variance can be high**.
- Actor–Critic + TD / GAE: Uses bootstrapping to reduce sampling requirements and potentially **lower variance**, but may introduce **gradient bias**.
- Compatible Function Approximation: Uses specific structural constraints and conditions. It can **have approximation errors** but the error won't bias the expected policy gradient.

Critic accuracy is not a necessary condition for Actor gradient unbiasedness. What truly matters is how Critic approximation errors enter the policy gradient estimator.

### RL Modules summary

| Module               | Core Question                                | Common Approaches                              |
| -------------------- | -------------------------------------------- | ---------------------------------------------- |
| Policy               | How to select an action?                     | Deterministic / Stochastic                     |
| Policy Optimization  | How to improve the policy?                   | Policy Gradient / Q-based Greedy Improvement   |
| Value Estimation     | How to estimate future returns?              | MC / TD / n-step TD                            |
| Advantage Estimation | How to evaluate a particular action?         | MC − Baseline / TD Error / GAE                 |
| Data Collection      | Where does the training data come from?      | On-policy / Off-policy                         |
| Environment Model    | Should we learn how the environment evolves? | Model-free / Model-based                       |
| Exploration          | How to explore unknown actions?              | ε-greedy / Entropy Bonus, etc.                 |
| Update Stability     | How to prevent training instability?         | Target Network / Clipping / Trust Region, etc. |

### Common algorithms & combination

| Algorithm | Core Components                                                    |
| --------- | ------------------------------------------------------------------ |
| REINFORCE | Stochastic Policy + Policy Gradient + Monte Carlo (MC)             |
| A2C       | Policy Gradient + Value Network + n-step Returns                   |
| PPO       | Policy Gradient + Value Network + Commonly GAE + Clipped Objective |
| DQN       | Q-learning + Neural Network + Experience Replay + Target Network   |

- n-step: observe n state transitions
- Timestep $t$: each transition +1
- episode: from start state to termination
