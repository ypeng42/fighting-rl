# Sutton (1999) vs. Konda & Tsitsiklis (1999)

Both papers study how a Critic can approximate the value function while still providing a correct Policy Gradient to the Actor.

## Key Differences

|                   | Sutton et al. (1999)                                                                                             | Konda & Tsitsiklis (1999)                                                                               |
| ----------------- | ---------------------------------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------- |
| Main Question     | Under what conditions does an approximate Critic produce the correct Policy Gradient?                            | Can Actor and Critic learn simultaneously without waiting for the Critic to fully converge?             |
| Critic            | Must satisfy two conditions: compatible function approximation and an appropriate error orthogonality condition. | Learns a projection of the Q-function onto features relevant to the Policy Gradient, using TD learning. |
| Actor Update      | Theoretical analysis assumes the Critic satisfies the required conditions when the Actor updates.                | Actor and Critic update simultaneously.                                                                 |
| Learning Strategy | Analyze the correctness of the gradient estimate.                                                                | Two-Time-Scale Learning: Critic learns faster; Actor changes more slowly.                               |
| Main Contribution | Establishes when an approximate Critic can provide an unbiased Policy Gradient.                                  | Establishes convergence guarantees for certain Actor-Critic algorithms with simultaneous learning.      |

## The Key Question Addressed by Konda & Tsitsiklis

Sutton's theoretical result assumes the Critic already satisfies the required conditions.

But in practice, the Critic is continuously learning.

What happens if we update the Actor **before the Critic has fully converged**?

Konda & Tsitsiklis address this using Two-Time-Scale Stochastic Approximation:

- Critic (fast): Learns to evaluate the Actor's current policy.
- Actor (slow): Updates its policy gradually, allowing the Critic to track the changing policy.

Under appropriate assumptions, this simultaneous learning process can converge to a stationary point.

## Shared Mathematical Core

Both papers are closely connected to the same fundamental condition:

\\[ \boxed{\mathbb E[\psi\_\theta(Q^\pi-\hat Q)]=0} \\]

where:

\\[ \psi\_\theta(s,a)=\nabla\_\theta\log\pi\_\theta(a\mid s) \\]

- \\(Q^\pi\\): True action-value function.
- \\(\hat Q\\): Critic's approximation.
- \\(\psi\_\theta\\): Policy score function.
- The expectation is taken under the relevant on-policy state-action distribution.

Key intuition: The Critic does not need to predict every Q-value perfectly. Its approximation errors must not distort the Policy Gradient.

- Sutton establishes conditions that make this possible.
- Konda & Tsitsiklis investigate how a TD-based Critic and Actor can learn together while preserving the convergence guarantees under suitable assumptions.

Even from a theoretical perspective, the two-time-scale condition is not a **universal requirement** for Actor-Critic methods.

The real question is:
>How much additional computational cost and slower learning are we willing to accept in exchange for a more reliable critic?
