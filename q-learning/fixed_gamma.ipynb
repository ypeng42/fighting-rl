## Why Does RL Usually Use a Fixed Discount Factor?

**Key idea:** A fixed discount factor $\gamma$ produces exponential discounting, which preserves **time consistency** and makes the Bellman equation simple.

### 1. Exponential Discounting: Time Consistency

With a fixed $\gamma$, the discount function is:

$$
D(k)=\gamma^k
$$

It satisfies:

$$
\boxed{D(k+m)=D(k)D(m)}
$$

This means preferences between two future rewards do not reverse merely because time passes.

### 2. Hyperbolic Discounting: Preference Reversal

Suppose we use a different discount function:

$$
D(k)=\frac{1}{1+k}
$$

We have two choices:

- **Option A:** Receive $10 in one step.
- **Option B:** Receive $18 in two steps.

**At time $t=0$:**

$$
V(A)=\frac{10}{2}=5
$$

$$
V(B)=\frac{18}{3}=6
$$

We prefer **B**.

**At time $t=1$:**

- Option A: Receive $10 immediately.
- Option B: Receive $18 in one step.

$$
V(A)=10
$$

$$
V(B)=\frac{18}{2}=9
$$

Now we prefer **A**!

Our preference reversed even though the rewards themselves did not change.

This is called **preference reversal**.

### 3. Why Does This Matter for RL?

With exponential discounting:

$$
V^\pi(s)=\mathbb{E}_\pi[R_{t+1}+\gamma V^\pi(S_{t+1})\mid S_t=s]
$$

The remaining future can be evaluated recursively using the same discount rule.

With hyperbolic discounting:

$$
D(k+m)\neq D(k)D(m)
$$

The discount applied to a future reward depends on how far away it is from the current decision time. This can create time-inconsistent preferences and complicate standard Bellman optimization.
