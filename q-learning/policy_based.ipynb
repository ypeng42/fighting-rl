minimize Loss or maximize J(total expected return)?

min L: use sample to make a Gradient Estimator.

Try to minizie:
> loss = -policy.log_prob(action) * reward

We know this will increase action's probablity (covered in llm series).
> min(-policy.log_prob(action))

So this means: the larger the reward, it will affect the P(action) accordingly! 
> min(-policy.log_prob(action) * reward)

Maximize J is what we want, min L helps us get there, it's called **surrogate loss**.

```python
import torch

theta = torch.tensor(0.0, requires_grad=True)
p = torch.sigmoid(theta)
J = p

true_gradient = torch.autograd.grad(J, theta, retain_graph=True)[0]
grad_if_forward = torch.autograd.grad(torch.log(p), theta)[0]

expected_gradient = p.item() * grad_if_forward.item()

print(f"P(Forward): {p.item():.2f}")
print(f"True dJ/dtheta: {true_gradient.item():.2f}")
print(f"Sampled gradient if Forward: {grad_if_forward.item():.2f}")
print(f"Expected sampled gradient: {expected_gradient:.2f}")
```
