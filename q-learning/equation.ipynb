# Bellman Equation & Q-learning

## 1. Discounted Return

The total discounted future reward:

$$
G_t = R_{t+1} + \gamma R_{t+2} + \gamma^2 R_{t+3} + \cdots
$$

Recursive form:

$$
G_t = R_{t+1} + \gamma G_{t+1}
$$

- $\gamma$: Discount factor, $0 \leq \gamma < 1$ for a continuing discounted task.
- $R_{t+1}$: Reward received after taking an action at time $t$.

## 2. Value Functions

State-value function:

$$
V^\pi(s) = \mathbb{E}_\pi[G_t \mid S_t=s]
$$

Action-value function:

$$
Q^\pi(s,a) = \mathbb{E}_\pi[G_t \mid S_t=s, A_t=a]
$$

- $V^\pi(s)$: Expected future return starting from state $s$ under policy $\pi$.
- $Q^\pi(s,a)$: Expected future return after taking action $a$ in state $s$, then following policy $\pi$.

## 3. Bellman Expectation Equations

State-value:

$$
V^\pi(s) =
\sum_a \pi(a \mid s)
\sum_{s',r} p(s',r \mid s,a)
\left[r + \gamma V^\pi(s')\right]
$$

Action-value:

$$
Q^\pi(s,a) =
\sum_{s',r} p(s',r \mid s,a)
\left[
r + \gamma \sum_{a'} \pi(a' \mid s')Q^\pi(s',a')
\right]
$$

These equations evaluate a fixed policy $\pi$.

## 4. Bellman Optimality Equations

Optimal state-value:

$$
V^*(s) =
\max_a \sum_{s',r} p(s',r \mid s,a)
\left[r + \gamma V^*(s')\right]
$$

Optimal action-value:

$$
\boxed{
Q^*(s,a) =
\sum_{s',r} p(s',r \mid s,a)
\left[r + \gamma \max_{a'} Q^*(s',a')\right]
}
$$

These equations describe the values of an optimal policy.

## 5. Q-learning Update

$$
\boxed{
Q(s,a) \leftarrow Q(s,a) +
\alpha\left[
r + \gamma \max_{a'} Q(s',a') - Q(s,a)
\right]
}
$$

TD Target (non-terminal transition):

$$
y = r + \gamma \max_{a'}Q(s',a')
$$

TD Error:

$$
\delta = y - Q(s,a)
$$

Update:

$$
Q_{\text{new}}(s,a) =
Q_{\text{old}}(s,a) + \alpha\delta
$$

For terminal transitions, set the future-value term to zero:

$$
y = r
$$

- $\alpha$: Learning rate.
- $\gamma$: Discount factor.
- $s'$: Next state.
- $a'$: Candidate action in the next state.

## 6. Relationship

**Bellman Optimality Equation** describes the mathematical relationship that the true optimal action values must satisfy.

**Q-learning** uses sampled transitions $(s,a,r,s')$ and current value estimates to iteratively approximate those optimal values.

Bellman Optimality Equation:

$$
Q^*(s,a) =
\mathbb{E}\left[
r + \gamma \max_{a'}Q^*(s',a')
\mid s,a
\right]
$$

Q-learning:

$$
Q(s,a) \leftarrow Q(s,a) +
\alpha\left[
r+\gamma\max_{a'}Q(s',a')-Q(s,a)
\right]
$$

**Key distinction:**

- Bellman Optimality Equation: expected value over possible outcomes, using $Q^*$.
- Q-learning: update using one observed outcome and current estimates $Q$.

$$
\pi^*(s) \in \arg\max_a Q^*(s,a)
